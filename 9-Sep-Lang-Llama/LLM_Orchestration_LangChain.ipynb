```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
import os

load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

print("LLM setup successful!")
```

```python
response = llm.invoke("Explain Machine Learning in 2 simple lines.")

print(response.content)
```

```python
prompt1 = ChatPromptTemplate.from_template(
    "Explain {topic} in simple language for a beginner."
)
```

```python
prompt2 = ChatPromptTemplate.from_template(
    "Convert the following explanation into 3 short bullet points:\n\n{explanation}"
)
```

```python
chain1 = prompt1 | llm

chain2 = prompt2 | llm
```

```python
from langchain_core.runnables import RunnablePassthrough

full_chain = (
    {"explanation": chain1}
    | chain2
)
```

```python
result = full_chain.invoke({
    "topic": "Machine Learning"
})

print(result.content)
```

```python
from langchain_core.tools import tool

@tool
def calculator(expression: str) -> str:
    """Calculate a mathematical expression."""
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except Exception:
        return "Invalid mathematical expression"

print(calculator.invoke("125 * 48"))
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[calculator]
)
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is 125 * 48?"
        }
    ]
})

print(response["messages"][-1].content)
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Explain what Machine Learning is in simple language."
        }
    ]
})

print(response["messages"][-1].content)
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is Generative AI?"
        }
    ]
})

print(response["messages"][-1].content)
```
