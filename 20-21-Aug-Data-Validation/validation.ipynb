```python
schema = pa.DataFrameSchema({
    "employee_id": Column(
        int,
        Check.greater_than(0),
        nullable=False,
        unique=True
    ),

    "employee_name": Column(
        str,
        nullable=False
    ),

    "email": Column(
        str,
        nullable=False,
        checks=Check.str_matches(
            r"^[\w\.-]+@[\w\.-]+\.\w+$"
        )
    ),

    "age": Column(
        int,
        Check.in_range(18, 65),
        nullable=False
    ),

    "department": Column(
        str,
        Check.isin(["IT", "HR", "Finance", "Marketing"]),
        nullable=False
    ),

    "salary": Column(
        int,
        Check.greater_than(0),
        nullable=False
    ),

    "joining_date": Column(
        str,
        nullable=False,
        checks=Check.str_matches(
            r"^\d{4}-\d{2}-\d{2}$"
        )
    ),

    "status": Column(
        str,
        Check.isin(["Active", "Inactive"]),
        nullable=False
    )
})
df
```

```python
try:
    validated_df = schema.validate(df, lazy=True)

    print("Validation Passed")
    df = validated_df

except pa.errors.SchemaErrors as e:

    print("Validation Failed")

    errors = e.failure_cases

    # Log validation errors
    logger.error("Data validation failed")

    for _, row in errors.iterrows():
        logger.error(
            f"Column: {row['column']} | "
            f"Check: {row['check']} | "
            f"Value: {row['failure_case']} | "
            f"Index: {row['index']}"
        )

    # Get invalid rows
    invalid_indexes = errors["index"].dropna().unique()

    # Keep only valid rows
    df = df.drop(index=invalid_indexes)

    print("Invalid rows:", len(invalid_indexes))
    print("Valid Data:")
    print(df)
```
