```python
from google import genai
from dotenv import load_dotenv
import os

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

chat = client.chats.create(
    model="gemini-3.6-flash"
)

response = chat.send_message(
    message="Hello! Introduce yourself in one sentence."
)

print(response.text)
```

```python
SYSTEM_PROMPT = """
You are a helpful AI assistant.

Rules:
- Answer clearly and accurately.
- Keep explanations simple and easy to understand.
- Maintain context from previous messages.
- If you don't know something, clearly say that you don't know.
- Do not invent or make up facts.
- Stay within your defined knowledge boundaries.
- Be polite and professional.
"""
```

```python
from google.genai import types

chat = client.chats.create(
    model="gemini-3.6-flash",
    config=types.GenerateContentConfig(
        system_instruction=SYSTEM_PROMPT
    )
)

response = chat.send_message(
    message="Explain what Python is in simple words."
)

print(response.text)
```
