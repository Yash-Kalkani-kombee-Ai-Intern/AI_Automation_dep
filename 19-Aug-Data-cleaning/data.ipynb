IQR = Q3 - Q1

Lower Bound = Q1 - (1.5 × IQR)

Upper Bound = Q3 + (1.5 × IQR)

---

Q1 → 25th percentile
Q2 → 50th percentile (Median)
Q3 → 75th percentile

```python
q1 = 41000.000000
q3 = 68500.000000

IQR = q3 - q1

Imin = q1 - (1.5 * IQR)
Imax = q3 + (1.5 * IQR)
df[(df["Salary"] < Imin) | (df["Salary"] > Imax)]
```

```python
df = df[(df["Salary"] >= Imin) & (df["Salary"] <= Imax)]
df
```
