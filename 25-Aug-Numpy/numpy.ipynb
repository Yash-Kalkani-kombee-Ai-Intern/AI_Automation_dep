```python
products = np.array(["Pen", "Book", "Bag", "Bottle"])
quantity = np.array([10, 5, 3, 8])
price = np.array([20, 100, 500, 150])

revenue = quantity * price

print("Products:", products)
print("Quantity:", quantity)
print("Price:", price)
print("Revenue:", revenue)

print("\nTotal sales:", np.sum(revenue))
print("Highest product revenue:", np.max(revenue))

# Products generating at least ₹500 revenue
mask = revenue >= 500
print("\nProducts with revenue >= ₹500:", products[mask])
print("Their revenue:", revenue[mask])
```
