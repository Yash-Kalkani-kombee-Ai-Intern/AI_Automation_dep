```python
import asyncio
import time
```

```python
async def io_task(name, delay):
    print(f"{name} started")

    await asyncio.sleep(delay)

    print(f"{name} finished")

    return f"{name} completed"
```

```python
async def main():

    start = time.perf_counter()

    results = await asyncio.gather(
        io_task("Task A", 2),
        io_task("Task B", 2),
        io_task("Task C", 2)
    )

    elapsed = time.perf_counter() - start

    print("\nResults:")
    print(results)

    print(f"\nTotal time: {elapsed:.2f} seconds")
```

```python
await main()
```

```python
import time
import asyncio
```

```python
def sync_io_task(name, delay):
    time.sleep(delay)
    return f"{name} completed"
```

```python
start = time.perf_counter()

sync_results = [
    sync_io_task("Task A", 2),
    sync_io_task("Task B", 2),
    sync_io_task("Task C", 2),
]

sync_time = time.perf_counter() - start

print("Results:")
print(sync_results)
print(f"\nSync time: {sync_time:.2f} seconds")
```

```python
async def async_io_task(name, delay):
    await asyncio.sleep(delay)
    return f"{name} completed"
```

```python
start = time.perf_counter()

async_results = await asyncio.gather(
    async_io_task("Task A", 2),
    async_io_task("Task B", 2),
    async_io_task("Task C", 2),
)

async_time = time.perf_counter() - start

print("Results:")
print(async_results)
print(f"\nAsync time: {async_time:.2f} seconds")
```

```python
improvement = ((sync_time - async_time) / sync_time) * 100
speedup = sync_time / async_time

print(f"Sync time       : {sync_time:.2f} sec")
print(f"Async time      : {async_time:.2f} sec")
print(f"Improvement     : {improvement:.2f}%")
print(f"Speedup         : {speedup:.2f}x")
```
