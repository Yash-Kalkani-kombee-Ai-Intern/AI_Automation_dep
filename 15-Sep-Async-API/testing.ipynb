```python
import asyncio
```

```python
async def risky_task(name, should_fail=False):
    print(f"{name} started")

    await asyncio.sleep(1)

    if should_fail:
        raise ValueError(f"{name} failed!")

    print(f"{name} finished")
    return f"{name} completed"
```

```python
async def exception_demo():
    try:
        result = await risky_task("Task A", should_fail=True)
        print(result)

    except ValueError as e:
        print(f"Error handled: {e}")


await exception_demo()
```

```python
async def slow_task():
    print("Slow task started")

    await asyncio.sleep(5)

    return "Slow task completed"
```

```python
async def timeout_demo():
    try:
        result = await asyncio.wait_for(
            slow_task(),
            timeout=2
        )

        print(result)

    except asyncio.TimeoutError:
        print("Task timed out!")


await timeout_demo()
```

```python
async def long_task():
    try:
        print("Long task started")

        await asyncio.sleep(10)

        print("Long task completed")
        return "Completed"

    except asyncio.CancelledError:
        print("Long task was cancelled")
        raise
```

```python
async def cancellation_demo():
    task = asyncio.create_task(long_task())

    await asyncio.sleep(2)

    print("Cancelling task...")
    task.cancel()

    try:
        await task
    except asyncio.CancelledError:
        print("Cancellation handled")


await cancellation_demo()
```
