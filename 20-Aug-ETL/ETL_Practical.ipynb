```python
std = 12356.104207

min = (df["salary"].mean()) - 3*std
max = (df["salary"].mean()) + 3*std

df = df[(df["salary"]>=min) & (df["salary"]<=max)]
df
```

```python
df["age"] = df["age"].fillna(df["age"].mode()[0])
df["joining_date"].unique()
df["joining_date"] = df["joining_date"].fillna(df["joining_date"].mode()[0])
df["city"] = df["city"].fillna(df["city"].mode()[0])
df = pd.get_dummies(df, columns=["city"], dtype=int)
df = pd.get_dummies(df, columns=["location"], dtype=int)
```
