```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3],
    "employee": ["Amit", "Neha", "Raj"],
    "dept_id": ["D01", "D02", "D01"],
    "salary": [40000, 50000, 35000]
})

departments = pd.DataFrame({
    "dept_id": ["D01", "D02"],
    "department": ["IT", "HR"]
})
```

```python
result = pd.merge(
    employees,
    departments,
    on="dept_id",
    how="left"
)

print(result)
```
