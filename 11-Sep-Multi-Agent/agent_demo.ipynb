```python
import os
from crewai import Agent, LLM

gemini_llm = LLM(
    model="gemini/gemini-3.6-flash",
    api_key=os.environ["GEMINI_API_KEY"]
)

research_agent = Agent(
    role="AI Researcher",
    goal="Research a topic and collect accurate, useful information",
    backstory="""
    You are an experienced AI researcher.
    You carefully analyze information and provide
    clear and useful findings for another agent.
    """,
    llm=gemini_llm,
    verbose=True
)

print("Research Agent + Gemini connected successfully ✅")
```

```python
from crewai import Task

research_task = Task(
    description="""
    Research the topic: "How Generative AI is changing businesses".

    Find the most important points about:
    1. Business use cases
    2. Benefits
    3. Risks
    4. Future impact

    Give clear and useful information that another agent
    can use to write a final report.
    """,
    expected_output="""
    A structured research summary containing:
    - Key business use cases
    - Major benefits
    - Important risks
    - Future impact
    """,
    agent=research_agent
)

print("Research Task created successfully ✅")
```

```python
from crewai import Crew, Process

research_crew = Crew(
    agents=[research_agent],
    tasks=[research_task],
    process=Process.sequential,
    verbose=True
)

result = await research_crew.kickoff_async()

print("\n===== RESEARCH RESULT =====\n")
print(result)
```

```python
from crewai import Agent, Task, Crew, Process

writer_agent = Agent(
    role="Professional Report Writer",
    goal="Turn research information into a clear, well-structured business report",
    backstory="""
    You are an experienced business report writer.
    You take research from another specialist and transform it
    into a professional and easy-to-understand report.

    You must only use information provided by the research.
    Do not invent facts.
    """,
    llm=gemini_llm,
    verbose=True
)
```

```python
writing_task = Task(
    description="""
    Using the research provided by the Research Agent,
    write a professional report on:

    "How Generative AI is changing businesses"

    The report should contain:
    1. Introduction
    2. Business use cases
    3. Benefits
    4. Risks
    5. Future impact
    6. Conclusion
    """,

    expected_output="""
    A clear professional business report with headings,
    based only on the research provided by the Research Agent.
    """,

    agent=writer_agent,
    context=[research_task]
)
```

```python
crew = Crew(
    agents=[
        research_agent,
        writer_agent
    ],

    tasks=[
        research_task,
        writing_task
    ],

    process=Process.sequential,
    verbose=True
)
```

```python
result = await crew.kickoff_async()

print("\n===== FINAL REPORT =====\n")
print(result)
```

```python
from crewai.tools import tool


@tool("Business Research Tool")
def business_research(topic: str) -> str:
    """
    Provides structured research information about a business-related topic.
    """

    knowledge_base = {
        "generative ai": """
        Generative AI can help businesses with:
        - Customer support automation
        - Software development
        - Content generation
        - Data analysis
        - Product development

        Major benefits include:
        - Increased productivity
        - Faster development
        - Cost reduction
        - Better customer personalization

        Important risks include:
        - Incorrect AI-generated information
        - Data privacy issues
        - Copyright concerns
        - Bias
        - Workforce impact
        """
    }

    topic_lower = topic.lower()

    for key, information in knowledge_base.items():
        if key in topic_lower:
            return information

    return "No research information found for this topic."
```

```python
research_agent = Agent(
    role="AI Researcher",

    goal="Research a topic and collect accurate, useful information",

    backstory="""
    You are an experienced AI researcher.
    You carefully analyze information and provide
    clear and useful findings for another agent.
    """,

    llm=gemini_llm,

    tools=[business_research],

    verbose=True
)
```

```python
from crewai.tools import tool


@tool("Report Validator Tool")
def validate_report(report: str) -> str:
    """
    Checks whether the generated report contains all required sections.
    """

    required_sections = [
        "Introduction",
        "Business Use Cases",
        "Benefits",
        "Risks",
        "Future Impact",
        "Conclusion"
    ]

    missing_sections = []

    for section in required_sections:
        if section.lower() not in report.lower():
            missing_sections.append(section)

    if not missing_sections:
        return "Report validation successful. All required sections are present."

    return f"Report is missing these sections: {', '.join(missing_sections)}"
```

```python
writer_agent = Agent(
    role="Professional Report Writer",

    goal="Turn research information into a clear, well-structured business report",

    backstory="""
    You are an experienced business report writer.
    You take research from another specialist and transform it
    into a professional and easy-to-understand report.

    You must only use information provided by the research.
    Do not invent facts.
    """,

    llm=gemini_llm,

    tools=[validate_report],

    verbose=True
)
```

```python
crew = Crew(
    agents=[
        research_agent,
        writer_agent
    ],
    tasks=[
        research_task,
        writing_task
    ],
    process=Process.sequential,
    verbose=True
)
```

```python
result = await crew.kickoff_async()

print("\n===== FINAL REPORT =====\n")
print(result)
```

```python
# Human-in-the-loop approval

print("\n===== GENERATED REPORT =====\n")
print(result)

print("\n" + "=" * 60)
approval = input("Do you approve this report? (approve/revise): ").strip().lower()

if approval == "approve":
    print("\n✅ Report approved by user.")
else:
    print("\n✏️ Report needs revision.")
    feedback = input("Enter your revision feedback: ")

    print("\nUser feedback:")
    print(feedback)
```

```python
# Human-in-the-loop: Revise the report

if approval == "approve":
    final_report = result
    print("\n✅ Final report approved.")

else:
    revision_task = Task(
        description=f"""
        Revise the previously generated report based on the user's feedback.

        ORIGINAL REPORT:
        {result}

        USER FEEDBACK:
        {feedback}

        Update the report according to the feedback.

        Keep these sections:
        1. Introduction
        2. Business Use Cases
        3. Benefits
        4. Risks
        5. Future Impact
        6. Conclusion
        """,
        expected_output="""
        A revised and professional business report
        that addresses the user's feedback.
        """,
        agent=writer_agent
    )

    revision_crew = Crew(
        agents=[writer_agent],
        tasks=[revision_task],
        process=Process.sequential,
        verbose=True
    )

    revised_result = await revision_crew.kickoff_async()

    final_report = revised_result

    print("\n===== REVISED FINAL REPORT =====\n")
    print(final_report)
```
