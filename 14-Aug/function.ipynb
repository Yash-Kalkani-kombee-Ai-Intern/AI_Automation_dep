```python
print("Enter 1 for Addition")
print("Enter 2 for Substraction")
print("Enter 3 for Multipilcation")
print("Enter 4 for Division")
number = int(input("Enter number for caclulation"))

num1 = float(input("Enter Number 1 = "))
num2 = float(input("Enter Number 2 = "))

def add(num1 , num2):
    result = num1 + num2
    return result

def div(num1 , num2):
    result = num1 / num2
    return result

def mul(num1 , num2):
    result = num1 * num2
    return result

def sub(num1 , num2):
    result = num1 - num2
    return result

if number == 1:
    print(add(num1 , num2))
elif number == 2:
    print(sub(num1 , num2))
elif number == 3:
    print(mul(num1 , num2))
elif number == 4:
    print(div(num1 , num2))

```

```python
def calculate(a, b):
    return a + b, a - b, a * b

add, sub, mul = calculate(10, 5)

print(add)
print(sub)
print(mul)
```

```python
def add_numbers(*numbers):
    total = 0

    for number in numbers:
        total += number

    return total

print(add_numbers(10, 20))
print(add_numbers(10, 20, 30, 40))
```
