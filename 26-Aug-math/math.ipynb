```python
total_students = 100
passed_students = 70

p_pass = passed_students / total_students

print("Probability of passing:", p_pass)
```

```python
students = pd.DataFrame({
    "gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "passed": [1, 1, 0, 1, 1, 0, 0, 1]
})

girls = students[students["gender"] == "F"]

p_pass_given_female = girls["passed"].mean()

print("P(Pass | Female):", p_pass_given_female)
```
